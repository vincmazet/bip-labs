# BIP Labs

This is the lab assignments (problems and programming exercises)
for the course [Basics of Image Processing](https://vincmazet.github.io/bip/)
in the [master IRIV](https://www.master-iriv.fr/) program at the [University of Strasbourg](https://www.unistra.fr/).

The programming exercises require Python, a free language used by many communities such as medicine, astronomy, and biology.
You will write your code in files called *notebooks* which combine code, images and formatted text.
In class, you will use either Jupyter Lab or VSCodium to write notebooks.
Please follow the instructions provided in @P:python.

Additional course materials, including supplementary resources, are available on the dedicated [Moodle page](https://moodle.unistra.fr/course/view.php?id=6678) (access restricted to my students).

After completing each exercise, you can refer to the correction below.
Furthermore, do not hesitate to ask your colleagues and teachers for additional information
or to discuss specific topics.
